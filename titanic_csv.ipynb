```python
import pandas as pd
#Part (c): Nominal Dissimilarity
# Define the URL for the Titanic dataset
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)

df["Sex"] = df["Sex"].map({"male":0,"female":1})

df = pd.get_dummies(df, columns=["Embarked"], drop_first=True)

r1 = df.loc[0, ["Sex","Embarked_Q","Embarked_S"]]
r2 = df.loc[1, ["Sex","Embarked_Q","Embarked_S"]]

dissimilarity = (r1 != r2).sum()/len(r1)

print("Nominal Dissimilarity:", dissimilarity)
```

```python
from math import sqrt
#Part (d): Numeric Dissimilarity
temp = df[["Age","Fare"]].dropna()

r1 = temp.iloc[0]
r2 = temp.iloc[1]

distance = sqrt(
    (r1["Age"]-r2["Age"])**2 +
    (r1["Fare"]-r2["Fare"])**2
)

print("Numeric Dissimilarity:", distance)
```
