```python
df['Age'] = df['Age'].fillna(df['Age'].median())
```

```python
df.drop_duplicates(inplace=True)

print("Shape after removing duplicates:")
print(df.shape)
```

```python
df['Sex'] = df['Sex'].map({'male':0,'female':1})

print("Unique values in Sex:")
display(df['Sex'].unique())
```

```python
import pandas as pd

# Load Titanic dataset
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)
```

```python
# Display first 5 rows
print(df.head())

# Dataset information
df.info()

# Missing values
print(df.isnull().sum())

# Statistical summary
print(df.describe())
```

```python
numeric_df = df.select_dtypes(include=['number'])
```

```python
print("Range")
print(numeric_df.max() - numeric_df.min())
```

```python
print("\nVariance")
print(numeric_df.var())
```

```python
print("\nStandard Deviation")
print(numeric_df.std())
```

```python
Q1 = numeric_df.quantile(0.25)
Q3 = numeric_df.quantile(0.75)

IQR = Q3 - Q1

print("\nInterquartile Range (IQR)")
print(IQR)
```

```python
dispersion = pd.DataFrame({
    "Range": numeric_df.max() - numeric_df.min(),
    "Variance": numeric_df.var(),
    "Standard Deviation": numeric_df.std(),
    "IQR": IQR
})

print(dispersion)
```

```python
print("\nConclusion")
print("Measures of dispersion such as Range, Variance, Standard Deviation, and IQR were successfully calculated for the Titanic dataset.")
```
